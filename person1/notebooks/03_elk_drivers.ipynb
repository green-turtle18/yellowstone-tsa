## Plain-English summary

These associations are provisional. The forecast evaluation is conditional retrospective prediction, not an operational forecast: harvest and full winter severity are not demonstrably available at the origin. Scoring excludes imputed targets and origins and unresolved survey coverage. Training uses observed elk transitions and past-only exogenous fills; transformations are re-estimated inside each origin. Q1 full-sample results deliberately retain flagged values for comparison with sensitivity checks. Seven screening tests use Bonferroni ×7, so the expected number of false positives under seven true nulls is 0.35 at 5%. A non-significant interaction cannot distinguish additive from compensatory mortality. ADF/KPSS use the longest contiguous observed segment rather than compressing calendar gaps. Q6 published hunting removals match late hunt, not the fall-plus-late totals used by the regression. The separate 0.26 adjustment is not total wolf kills. Q6 uses level harvest rate as specified, even if its stationarity is uncertain; see Q1 for the differenced alternative. A coefficient sign does not prove the mortality mechanism.
